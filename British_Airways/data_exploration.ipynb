Get the dataset - Countries

```python
df = pd.read_csv('ba_reviews.csv')
df.info()
```

```python
df.duplicated().sum()
```

```python
df.head()
```

```python
import matplotlib.pyplot as plt

# create box plot of rating by year. Only date column i have in dataset. so need to extract year from date
df['Year'] = pd.to_datetime(df['date'], dayfirst=True).dt.year
plt.figure(figsize=(10,6))
df.boxplot(column='rating', by='Year')
plt.suptitle('')   # removes automatic pandas title
plt.ylabel('Rating')
plt.xticks(rotation=45)
plt.show()
```
