# Data Cleaning and Preprocessing

```python
# In ride-sharing data, nulls often aren't "missing" information; they are "conditional." '
# 'For example, if a ride was canceled, it won't have a Driver_Rating or/ Payment_Method
```

# Handle missing values
Instead of dropping rows, we will fill them with logical defaults so the data remains compatible with MySQL

```python
# Fill rating with 0 or/ placeholder since "0" usually indicates no rating given
df_ola_ride['Driver_Ratings'] = df_ola_ride['Driver_Ratings'].fillna(0) 
df_ola_ride['Customer_Rating'] = df_ola_ride['Customer_Rating'].fillna(0)
```

```python
df_ola_ride['Driver_Ratings'].isnull().sum()  
```

```python
# Fill TAT (Turn Around Time) with the 0
df_ola_ride['C_TAT'] = df_ola_ride['C_TAT'].fillna(0)
df_ola_ride['V_TAT'] = df_ola_ride['V_TAT'].fillna(0)
```
