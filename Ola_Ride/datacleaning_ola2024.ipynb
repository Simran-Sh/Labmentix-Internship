# The Strategic Mapping Logic - Numerucal columns

```python
# If successful booking, fill NaN ratings with 0, else for the cancelled ones, fill with -1 for non-applicable)

df_ola['Driver_Ratings'] = np.where(
                                    df_ola['Booking_Status'] == 'Success', 
                                           df_ola['Driver_Ratings'].fillna(0),-1)
                                            
                                            
df_ola['Customer_Rating'] = np.where(
                                    df_ola['Booking_Status'] == 'Success',
                                          df_ola['Customer_Rating'].fillna(0),-1)  

```

```python
#  Fill nulls with 0

df_ola['V_TAT'] = df_ola['V_TAT'].fillna(0)
df_ola['C_TAT'] = df_ola['C_TAT'].fillna(0)

```

# Handling Catgorical columns

```python
df_ola['Incomplete_Rides'].value_counts()
```

# 3,926 "Yes"
# 60,041 "No"
# The rest (39k) are Null (Canceled)

```python
df_ola['Incomplete_Rides_Reason'].value_counts()
```

```python
df_ola['Canceled_Rides_by_Customer'] = df_ola['Canceled_Rides_by_Customer'].fillna('Not applicable') 
# this column having null values means ride was not cancelled by customer

df_ola['Canceled_Rides_by_Driver'] = df_ola['Canceled_Rides_by_Driver'].fillna('Not applicable') 
# this column having null values means ride was not cancelled by driver

df_ola['Incomplete_Rides'] = df_ola['Incomplete_Rides'].fillna('N/A - Never Started')
df_ola['Incomplete_Rides_Reason'] = df_ola['Incomplete_Rides_Reason'].fillna('N/A - Completed or Cancelled')


```

```python
df_ola['Payment_Method'] = df_ola['Payment_Method'].fillna('No Payment')
```
